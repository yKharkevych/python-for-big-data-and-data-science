6. (7б) Визначте функцію, яка приймає iterable (ітерабельний обʼєкт, напр. список), що складається з 7ми рядкових значень - назв погодних умов (напр. "sunny") за кожен день у тижні. Повертає функція словник, де ключами є рядки з погодними умовами, а значеннями - кількість разів, скільки дана погода була протягом тижня (напр. {'sunny': 1, ...}). Викличте цю функцію та виведіть результат її роботи на екран.

```python
def count_weather(weather):
  weather_counts = {}
  for condition in weather:
    if condition in weather_counts:
      weather_counts[condition]+=1
    else:
      weather_counts[condition]=1
  return weather_counts

w = ["sunny", "cloudy", "rainy", "sunny", "sunny", "cloudy", "stormy"]
res = count_weather(w)
print(res)
```

7. (15б) Напишіть програму, яка б шукала необхідне число у впорядкованому списку цілочислених значень, де
число має вводитись з консолі.
власне алгоритм пошуку винесений в окрему функцію, яка повертає булеве значення (знайдено число - True, не знайдено - False).
в залежності від результату функції, виведіть на екран повідомлення про успішний або ж неуспішний пошук.
Зауваження: використовувати оператор in заборонено.

```python
def search_num(n, n_list):
  lo = 0
  hi = len(n_list) - 1
  while lo<=hi:
    mid = (lo+hi)//2
    guess = n_list[mid]
    if guess==n:
      return True
    if n>guess:
      lo = mid+1
    else:
      hi = mid-1
  return False

numbers = [2, 3, 5, 7, 10, 14, 15, 18, 26, 28, 29, 31, 35]
try:
  n = int(input("Enter number: "))
  if search_num(n, numbers) == True:
    print("number is found")
  else:
    print("number is not in the list")
except ValueError:
  print("print real number")
```

# III. Decorators.

11. (11б) Визначте декоратор, який би виводив у консоль повідомлення про початок роботи і назву функції (напр. my_function._ _ name _ _), а після відпрацювання функції - повідомлення про закінчення роботи і назву функції, що задекорована.

```python
def decorator(func):
  def wrapper(*args, **kwargs):
    print(f"start working: {func.__name__}")
    res = func(*args, **kwargs)
    print(f"finish: {func.__name__}")
    return res
  return wrapper

@decorator
def sum(a, b):
  return a+b

res = sum(1,2)
```

15. (15б) Визначте функцію, яка рекурсивно реалізовує пошук числа у послідовності Фібоначчі за її індексом (нумерація починається з нуля). Наприклад, потрібно знайти перше число у послідовності, тоді функція прийматиме 0 у якості аргумента і повертатиме 1, якщо ж знаходити третє число (індекс 2) послідовності, то функція повертатиме 2, якщо ж знаходити десяте число (індекс 9), то функція повертатиме 55.

```python
def fib(n):
  if n==0 or n==1:
    return 1
  return fib(n-1)+fib(n-2)

print(f"n = 0: {fib(0)}")
print(f"n = 1: {fib(1)}")
print(f"n = 9: {fib(9)}")
```
