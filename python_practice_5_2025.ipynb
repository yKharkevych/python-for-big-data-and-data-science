1. (6б). Створіть клас Building, додайте три атрибути, один з яких має значення за замовчуванням та два методи на свій розсуд.

```python
class Building:
    def __init__(self, name, year_built, floors = 1):
        self.name = name
        self.year_built = year_built
        self.floors = floors

    def __repr__(self):
        return f'{self.name}, {self.floors} floors, built in {self.year_built}'

    def add_floors(self, count):
        if count > 0:
            self.floors += count
            print(f'{self.name} now has {self.floors} floors.')
        else:
            print('Floors cannot be negative.')

my_house = Building('My House', 1970)
print(my_house)
my_house.add_floors(2)
```

6. (11б). Створіть 2 дочірніх класи від 3d-фігура: Циліндр та Куб, які мають атрибути, необхідні для розрахунку об'єму. Визначте метод get_volume в кожному з них так, щоби вони розраховували об'єм у залежності від формули для кожного типу фігури. Створіть обʼєкт класу Циліндр та обʼєкт класу Куб. Виконайте операцію суми за допомогою оператора + між ними, аби знайти сумарний об'єм.

```python
class Figure_3D:
    def get_volume(self):
        return 0

    def __add__(self, other):
        return self.get_volume() + other.get_volume()

class Cube(Figure_3D):
    def __init__(self, side):
        self.__side = side

    def get_volume(self):
        return self.__side ** 3

class Cylinder(Figure_3D):
    def __init__(self, radius, height):
        self.__radius = radius
        self.__height = height

    def get_volume(self):
        return self.__radius * self.__height * 3.14

c = Cube(5)
print(f"Volume of cube is {c.get_volume()}")
cylinder = Cylinder(5, 10)
print(f"Volume of cylinder is {cylinder.get_volume()}")
total_volume = c + cylinder
print(f"Total volume is {total_volume}")
```

7. (3б). Продемонструйте різницю між isinstance та issubclass.

```python
class Figure_3D:
    def get_volume(self):
        return 0

    def __add__(self, other):
        return self.get_volume() + other.get_volume()

class Cube(Figure_3D):
    def __init__(self, side):
        self.__side = side

    def get_volume(self):
        return self.__side ** 3

class Cylinder(Figure_3D):
    def __init__(self, radius, height):
        self.__radius = radius
        self.__height = height

    def get_volume(self):
        return self.__radius * self.__height * 3.14

c = Cube(5)
cylinder = Cylinder(5, 10)

# перевіряє чи є конкретний екземпляр представником певного класу
print(isinstance(c, Cube))
print(isinstance(cylinder, Figure_3D))
#перевіряє чи є клас розширенням іншого
print(issubclass(Cube, Figure_3D))
```

8. (13б). Створіть клас EWallet з приватними атрибутами balance та id. Реалізуйте методи поповнення та зняття коштів, забезпечивши належну інкапсуляцію. Підказка: використовуйте декоратори property та setter.

```python
class EWallet:
    def __init__(self, id, balance):
        self.__balance = balance
        self.__id = id

    @property
    def balance(self):
        return self.__balance

    @property
    def id(self):
        return self.__id

    @balance.setter
    def balance(self, value: float):
        if value < 0:
            self.__balance = 0
        else:
            self.__balance = value

    def deposit(self, amount: float):
        if amount >= 0:
            self.__balance += amount
            print(f"Рахунок поповнено на {amount}. Поточний баланс: {self.balance}")

    def withdraw(self, amount: float):
        if amount > 0:
            self.__balance -= amount
            print(f"Знято {amount}. Залишок: {self.balance}")

my_wallet = EWallet("A777", 100)

print(f" ID: {my_wallet.id}")
my_wallet.deposit(50)
my_wallet.withdraw(30)
my_wallet.balance = 10
print(f" Новий баланс: {my_wallet.id}")
```

9. (11б). Створіть клас ShoppingCart, який містить список об'єктів типу Item. Реалізуйте методи для додавання айтемів, їх видалення та відображення списку айтемів.

```python
class Item:
    def __init__(self, name):
        self._name = name

    def __eq__(self, other):
        if not isinstance(other, Item):
            return False
        return self._name == other._name

    def __repr__(self):
        return self._name

class ShoppingCard:
    def __init__(self, items=None):
        if items is None:
            self.items = []
        else:
            self.items = items

    def add_item(self, item):
        self.items.append(item)

    def delete_item(self, item):
        self.items.remove(item)

    def __repr__(self):
        return f"ShoppingCard({self.items})"

sc = ShoppingCard()
sc.add_item(Item("apple"))
sc.add_item(Item("banana"))
sc.add_item(Item("orange"))
sc.add_item(Item("pear"))
print(sc)
sc.delete_item(Item("apple"))
sc.delete_item(Item("orange"))
print(sc)
```

Створіть клас Car, який успадковує обидва класи Vehicle та Engine. Продемонструйте використання множинної спадковості, створивши обʼєкт класу Car та отримавши mro для цього класу.

```python
class Vehicle:
    def __init__(self, brand, production_age):
        self.brand = brand
        self.production_age = production_age
        print("Init Vehicle")

    def vehicle_info(self):
        return f"brand: {self.brand}, {self.production_age} years old"

class Engine:
    def __init__(self, type, volume):
        self.type = type
        self.volume = volume
        print("Init Engine")

    def engine_info(self):
        return f"type: {self.type}, volume: {self.volume}"

class Car(Vehicle, Engine):
    def __init__(self, brand, production_age, type, volume, model_name):
        Vehicle.__init__(self, brand, production_age)
        Engine.__init__(self, type, volume)
        self.model_name = model_name
        print("Init Car")

    def car_info(self):
        return f"{self.model_name} | {self.vehicle_info()} | Engine: {self.engine_info()}"

my_car = Car("Toyota", 3, "Petrol", 2.5, "Camry")
print(my_car.car_info())
for c in Car.mro():
    print(c)
```

Ітерабельний об'єкт це такий об'єкт, котрий можна переглядати крок за кроком (строка str), а ітератор це ніби курсор, котрий пам'ятає, на якому кроці зараз знахолиться і виводить по одному елементу цього об'єкту (по одному символу)

13. (11б). Створіть клас, що має визначений Iterator Protocol та при кожному виклику next повертає назву дня тижня, поки вони не вичерпаються. Зверніть увагу, що порядок має значення.

```python
class Week:
    def __init__(self):
        self.days = ["Понеділок", "Вівторок", "Середа", "Четвер", "П'ятниця", "Субота", "Неділя"]
        self.i = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.i < len(self.days):
            d = self.days[self.i]
            self.i += 1
            return d
        else:
            raise StopIteration

week = Week()
for day in week:
    print(day)
```
